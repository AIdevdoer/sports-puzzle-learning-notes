# Sports Puzzle Practice Notebook

This notebook turns three sports-memory activities into short daily drills. It is designed for learners who want a repeatable way to notice player roles, team identity, and visual details.

## Drill 1: NFL player clues

Write down the position, conference, team colors, and one career clue before making a guess. Then compare your reasoning with the daily [NFL Wordle](https://wordlecup.today/en/nfl/).

## Drill 2: Football logo recall

Study shape, color, lettering, and regional symbols separately. Test the same cues with the [football logo memory game](https://wordlecup.today/en/football/).

## Drill 3: Basketball player profiles

Group clues by position, era, franchise, and nationality. Use the [basketball player puzzle](https://wordlecup.today/en/basketball-wordle/) for a six-guess practice round.


```python
drills = {
    "NFL player clues": ["position", "conference", "team colors", "career clue"],
    "Football logo recall": ["shape", "color", "lettering", "regional symbol"],
    "Basketball profiles": ["position", "era", "franchise", "nationality"],
}
for name, cues in drills.items():
    print(name + ": " + ", ".join(cues))
```

## Five-minute routine

1. Pick one drill.
2. Record the clues before guessing.
3. Explain why each clue changed your shortlist.
4. Save one mistake as tomorrow&apos;s review cue.

The aim is deliberate recall, not speed. A short written explanation makes the next attempt more useful.
